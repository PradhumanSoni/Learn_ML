**We will Learn How to Encode our Categorical Features and Target Variable**

```python
import pandas as pd
```

```python
df = pd.read_csv('customer.csv')
print(df.head())
```

### We will work on Only Ordinal and Label Encoding For now, later on we can update the file by using 1 hot encoding on Gender Feature and then finally using ColumnTransformer to build a pipeline to complete all encoding at once

```python
X = df.iloc[:,2:4]
y = df.iloc[:,-1]
```

```python
print(X['review'].value_counts())
```

```python
print(X['education'].value_counts())
```

- We have three types of reviews having order good > average > poor
- We have three categories of education order PG > UG > School

```python
# Let us train/test split from start itself to prevent Data Leakage
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.3 , random_state = 0)
```

```python
# Let us make OrdinalEncoder object
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder

oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])

# Let us fit and transform the Training Feature data
X_train_encoded = pd.DataFrame(oe.fit_transform(X_train) , columns = X_train.columns)
X_test_encoded = pd.DataFrame(oe.transform(X_test) , columns = X_test.columns)
```

```python
# Let us take a loot at it
print(f"Before Encoding -> \n{X_train['education'].head(10)}")
print(f"After Encoding -> \n{X_train_encoded['education'].head(10)}")
```

```python
print(y_train)
```

```python
# Now let us Label Encode the target Variable
le = LabelEncoder() # it does not need any parameter
y_train_encoded = pd.Series(le.fit_transform(y_train) , index = y_train.index)
y_test_encoded = pd.Series(le.transform(y_test) , index = y_test.index)
```

```python
print(y_train_encoded.head(10))
```

### Conclusion 
- We encoded the Ordinal Features using OrdinalEncoder
- We encoded the Target Variable using LabelEncoder
