```python
df = pd.read_csv('covid_toy.csv')
print(df.head())
```

```python
df.info()
```

- We see 10 missing values in fever column. So we will fill those using SimpleImputer just to go with the workflow

```python
# Let us check if outliers are there in Fever column
df['fever'].describe()
```

- There are No Outliers so we will Impute with mean fever only 

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,:-1], df.iloc[:,-1], test_size=.2 , random_state=0)
X_train.reset_index(drop=True,inplace=True)
X_test.reset_index(drop=True,inplace=True)
y_train.reset_index(drop=True , inplace=True)
y_test.reset_index(drop=True , inplace=True)
```

```python
print(f"X_train -> {X_train.shape}")
print(f"y_train -> {y_train.shape}")
print(f"X_test -> {X_test.shape}")
print(f"y_test -> {y_test.shape}")
```

```python
# Let us look at the Ordinal Feature and the flow of it
print(X_train['cough'].value_counts())
print(X_train['city'].value_counts())
```

- So we got two categories where Strong > Mild
- As for Nominal Categorical columns we don't need to specify the trend

### ***AAM ZINDAGI***

**Simple Imputing Fever Feature**

```python
# Simple Impute the fever column to fill the null values
si = SimpleImputer(strategy='mean') # we will impute with mean fever
X_train['fever'] = si.fit_transform(X_train[['fever']]) # fit transform for one column, we need to pass DataFrame
X_test['fever'] = si.transform(X_test[['fever']])
```

```python
print(X_train['fever'].isna().sum()) # See we have imputed the Nulls 
```

**Label Encoding Target Variable**

```python
# Using label Encoder On Target Variable
le = LabelEncoder()
y_train_encoded = pd.Series(le.fit_transform(y_train))
y_test_encoded = pd.Series(le.fit_transform(y_test))
```

**Ordinal Encoding Cough Feature**

```python
# Now let us Encode the Ordinal Columns
oe = OrdinalEncoder(categories=[['Mild','Strong']] , dtype=np.int32)
X_train_ord = pd.DataFrame(oe.fit_transform(X_train[['cough']]), columns=X_train[['cough']].columns)
X_test_ord = pd.DataFrame(oe.transform(X_test[['cough']]), columns=X_test[['cough']].columns)
```

```python
print(X_train_ord)
```

**Nominal Encoding (One Hot Encoding) Gender and City Columns**

```python
ohe = OneHotEncoder(drop='first' , sparse_output=False , dtype=np.int32)

X_train_ohe = pd.DataFrame(ohe.fit_transform(X_train[['gender','city']]), columns = ohe.get_feature_names_out())
X_test_ohe = pd.DataFrame(ohe.transform(X_test[['gender','city']]), columns = ohe.get_feature_names_out())
```

### ***Concatenating All the DataFrames With Original DataFrame***

```python
X_train_encode = pd.concat([X_train_ohe , X_train_ord] , axis=1)
X_test_encode = pd.concat([X_test_ohe , X_test_ord] , axis=1)
```

```python
final_train = pd.concat([X_train.drop(['gender','city','cough'],axis=1), X_train_encode],axis=1)
final_test = pd.concat([X_train.drop(['gender','city','cough'],axis=1), X_train_encode],axis=1)
```

```python
print(final_train)
```

### ***Thus we finally Encoded our all the columns (Such a hectic JOB to do )***

### ***MENTOS ZINDAGI***

```python
# Now We will work with ColumnTransformer class transformer
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(transformers=[
    ('impute_fever' , SimpleImputer(strategy='mean') , ['fever']),
    ('onehotencode' , OneHotEncoder(drop='first' , sparse_output=False , dtype=np.int32) , ['gender','city']),
    ('ordinal' , OrdinalEncoder(categories=[['Mild','Strong']] , dtype=np.int32) , ['cough'])
] , remainder= 'passthrough')
```

```python
transformer.set_output(transform='pandas')
X_train_transformed = transformer.fit_transform(X_train)
X_test_transformed = transformer.transform(X_test)
```
