```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from statsmodels.api import OLS
```

```python
X,y = load_diabetes(return_X_y= True)
```

```python
print(y.shape) # Series
```

```python
print(X.shape) # Dataframe with 10 features
```

```python
X = np.insert(X,0,1,axis=1) # This adds 1 to first column
```

```python
model = OLS(y,X)
```

```python
results = model.fit()
```

```python
print(results.summary())
```

```python
# --- Interpretation of some key results (explained below) ---
print("\n--- Key Statistics from the OLS Model ---")
print(f"R-squared: {results.rsquared:.4f}")
print(f"Adjusted R-squared: {results.rsquared_adj:.4f}")
print(f"F-statistic: {results.fvalue:.2f}")
print(f"Prob (F-statistic): {results.f_pvalue:.4f}")
print("\nCoefficients:")
print(results.params)
print("\nP-values for Coefficients:")
print(results.pvalues)
```
