```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from statsmodels.api import OLS
```

```python
X,y = load_diabetes(return_X_y = True)
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(2,5 , figsize=(20,10))
ax = ax.flatten()
for i in range(10):
    ax[i].scatter(X[:,i] , y)
    ax[i].set_title(f"plot {i+1}")
```

```python
# Checking linearity using Residual plots for diabetes dataset
lr = LinearRegression()
lr.fit(X,y)
```

```python
predicted = lr.predict(X)
```

```python
plt.scatter(predicted ,y - predicted , label = 'Residual PlotPoints')
plt.axhline(0, color='red' , linestyle = '--',label='Zero Line')
plt.xlabel("Predicted Values")
plt.ylabel("Residual Values")
plt.title("Residual Plot Against Predicted values")
plt.legend()
plt.show()
```

`We can see that the residual plot is randomly scattered around the line x = 0`                                           
`-> Thus we can say that the data may be linear -> Still cannot confirm`
