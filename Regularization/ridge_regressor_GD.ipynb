### ***Using Gradient Descent Algorithm Inside Ridge Regression***

```python
from sklearn.datasets import make_regression
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression , Ridge
from sklearn.metrics import r2_score
```

```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import mean_absolute_error , root_mean_squared_error , r2_score
from sklearn.model_selection import train_test_split
X,y = load_diabetes(return_X_y =True)

X_train,X_test,y_train,y_test = train_test_split(X,y , test_size = .2 , random_state = 4)

print(f"Shape X_train -> {X_train.shape}")
print(f"Shape X_test -> {X_test.shape}")
print(f"Shape y_train -> {y_train.shape}")
print(f"Shape y_test -> {y_test.shape}")
from sklearn.linear_model import Ridge

rr = Ridge(0.001 , solver = "sparse_cg" , max_iter = 500)  # sparse_cg used Gradient Method To solve Ridge Regression

rr.fit(X_train , y_train)
y_pred = rr.predict(X_test)

print(f"Intercept -> {rr.intercept_}")
print(f"Weightages -> {rr.coef_}")
print("\n")
print(f"R2 Score Test-> {r2_score(y_test, y_pred)}")
```

### ***Ridge Regressor -> OLS Method -> "Choslesky / Gradient Method -> sparse_cg"***

```python
class RidgeGD:
    def __init__(self,alpha=0.1 , epochs = 100 , learning_rate = 0.1):
        self.alpha = alpha
        self.m = None
        self.b = None
        self.epochs = epochs
        self.lr = learning_rate

    def fit(self, X_train , y_train):
        order = X_train.shape[1]
        coefs = np.ones(order+1)    # 
        coefs[0] = 0     
        X_train = np.insert(X_train , 0 , 1 ,axis = 1)                                                     
    
        for i in range(self.epochs):
            slope = (X_train.T @ X_train @ coefs) - (X_train.T @ y_train) + (self.alpha * coefs)
            coefs = coefs - self.lr * slope
        
        self.m = coefs[1:]
        self.b = coefs[0]

    def predict(self, X_test):
        return  X_test @ self.m + self.b
```

```python
ridge_gd = RidgeGD(alpha = 0.001 , epochs = 500 , learning_rate = 0.005)
ridge_gd.fit(X_train , y_train)

print(r2_score(y_test , ridge_gd.predict(X_test)))
print(ridge_gd.m)
print(ridge_gd.b)
```
