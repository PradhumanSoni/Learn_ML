```python
import pandas as pd

data = pd.read_csv('https://raw.githubusercontent.com/npradaschnor/Pima-Indians-Diabetes-Dataset/master/diabetes.csv')

data.head()
```

```python
X = data.drop('Outcome', axis=1)
y = data['Outcome']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
y_scores = model.predict_proba(X_test)[:,1]
```

```python
y_scores
```

```python
from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(y_test, y_scores)
```

```python
thresholds
```

```python
import plotly.graph_objects as go
import numpy as np


# Generate a trace for ROC curve
trace0 = go.Scatter(
    x=fpr,
    y=tpr,
    mode='lines',
    name='ROC curve'
)

# Only label every nth point to avoid cluttering
n = 10
indices = np.arange(len(thresholds)) % n == 0  # Choose indices where index mod n is 0

trace1 = go.Scatter(
    x=fpr[indices],
    y=tpr[indices],
    mode='markers+text',
    name='Threshold points',
    text=[f"Thr={thr:.2f}" for thr in thresholds[indices]],
    textposition='top center'
)


# Diagonal line
trace2 = go.Scatter(
    x=[0, 1],
    y=[0, 1],
    mode='lines',
    name='Random (Area = 0.5)',
    line=dict(dash='dash')
)

data = [trace0, trace1, trace2]

# Define layout with square aspect ratio
layout = go.Layout(
    title='Receiver Operating Characteristic',
    xaxis=dict(title='False Positive Rate'),
    yaxis=dict(title='True Positive Rate'),
    autosize=False,
    width=800,
    height=800,
    showlegend=False
)

# Define figure and add data
fig = go.Figure(data=data, layout=layout)

# Show figure
fig.show()
```

```python
# Assume that fpr, tpr, thresholds have already been calculated
optimal_idx = np.argmax(tpr - fpr)
optimal_threshold = thresholds[optimal_idx]
print("Optimal threshold is:", optimal_threshold)
```

```python
import numpy as np
import plotly.graph_objects as go
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import roc_curve, roc_auc_score
from sklearn.preprocessing import StandardScaler

# Assuming that X_train, X_test, y_train, y_test are already defined

# SVM requires feature scaling for better performance
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Logistic Regression model
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)
lr_scores = lr_model.predict_proba(X_test)[:,1]

# SVM model
svm_model = SVC(probability=True)
svm_model.fit(X_train_scaled, y_train)
svm_scores = svm_model.predict_proba(X_test_scaled)[:,1]

# Generate ROC curve data for logistic regression model
lr_fpr, lr_tpr, lr_thresholds = roc_curve(y_test, lr_scores)
lr_auc = roc_auc_score(y_test, lr_scores)

# Generate ROC curve data for SVM model
svm_fpr, svm_tpr, svm_thresholds = roc_curve(y_test, svm_scores)
svm_auc = roc_auc_score(y_test, svm_scores)

# Generate a trace for the Logistic Regression ROC curve
trace0 = go.Scatter(
    x=lr_fpr,
    y=lr_tpr,
    mode='lines',
    name=f'Logistic Regression (Area = {lr_auc:.2f})'
)

# Generate a trace for the SVM ROC curve
trace1 = go.Scatter(
    x=svm_fpr,
    y=svm_tpr,
    mode='lines',
    name=f'SVM (Area = {svm_auc:.2f})'
)

# Diagonal line
trace2 = go.Scatter(
    x=[0, 1],
    y=[0, 1],
    mode='lines',
    name='Random (Area = 0.5)',
    line=dict(dash='dash')
)

data = [trace0, trace1, trace2]

# Define layout with square aspect ratio
layout = go.Layout(
    title='Receiver Operating Characteristic',
    xaxis=dict(title='False Positive Rate'),
    yaxis=dict(title='True Positive Rate'),
    autosize=False,
    width=800,
    height=800,
    showlegend=True
)

# Define figure and add data
fig = go.Figure(data=data, layout=layout)

# Show figure
fig.show()
```

```python
import pandas as pd

data = pd.read_csv('diabetes.csv')

data.head()
```

### We will work on Indians Diabetes Sample Data

```python
print(data.info())

print(data.describe())
```

```python
X = data.iloc[:,:-1]
y = data.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2 , random_state=2)
print(f"Shape X_train - > {X_train.shape}")
print(f"Shape X_test -> {X_test.shape}")
print(f"Shape X_train - > {y_train.shape}")
print(f"Shape X_test -> {y_test.shape}")
```

```python
# Let us fit the Logit Model
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train,y_train)

y_pred = lr.predict(X_test)
y_prob = lr.predict_proba(X_test) # This gives up an array of shape (len , 2)
# (i) Having Probabilities of Negatives
# (ii) Having Probabilities of Positives
```

```python
print(y_prob.shape) # This shows the 2D array shape
```

```python
positive_prob = y_prob[:,1]
```

```python
# Now Let us check the roc_curve values i.e tpr , fpr and thresholds

from sklearn.metrics import roc_curve
fpr , tpr , thresholds = roc_curve(y_test , positive_prob)
```

```python
print(tpr) # These are the values of true postiive rate for each threshold formed
```

```python
print(fpr) # These are the values of false positive rate for each threshold
```

```python
print(thresholds) # These are the corresponding Threshold values
```

```python
# Let us plot the scatter
import matplotlib.pyplot as plt

plt.plot(fpr ,tpr)
```

```python
# Optimum Threshold?
import numpy as np
best_idx = np.argmax(tpr - fpr) # This will give the array again, the biggest value of that array holds the best threshold (since it is the optimal value)

best_threshold = thresholds[best_idx]

print(f"Best Threshold Value -> {best_threshold}")
```

```python
y_pred_best = [1 if i>=best_threshold else 0 for i in positive_prob ]
```

```python
from sklearn.metrics import accuracy_score
print(f"Accuracy Score (Threshold = 0.5) -> {accuracy_score(y_test , y_pred)}")
print(f"Accuracy Score (Threshold = 0.36) -> {accuracy_score(y_test , y_pred_best)}")
```

```python
from sklearn.metrics import confusion_matrix
print(f"Confusion Matrix (Threshold = 0.5) -> \n{confusion_matrix(y_test , y_pred)}\n")
print(f"Confusion Matrix (Threshold = 0.36) -> \n{confusion_matrix(y_test , y_pred_best)}\n")
```

```python
from sklearn.metrics import classification_report
print(f"Classification Report (Threshold = 0.5) -> \n{classification_report(y_test , y_pred)}\n")
print(f"Classification Report (Threshold = 0.36) -> \n{classification_report(y_test , y_pred_best)}\n")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('seaborn-v0_8-whitegrid')

df = sns.load_dataset('iris')
```

```python
print(df.head()) # This is our Iris dataset
```

### Basic Info about the data

```python
print(df.info())

print(df.describe())
```

### Looking at the Distribution of categories of species

```python
df['species'].value_counts().plot(kind='bar') # There is no diff in the distribution of each species
```

### First find the accuracy of Logistic Regression Model with original threshold = 0.5

```python
# Let us Label Encode the species (target variable) First
from sklearn.preprocessing import LabelEncoder
X = df.iloc[:,:4] # Separate Features 
y = df.iloc[:,-1] # separate target variable

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.4 , random_state=42)
label_encode = LabelEncoder()

y_train_enc = pd.Series(label_encode.fit_transform(y_train))
y_test_enc = pd.Series(label_encode.transform(y_test))

print(y_train_enc.sample(5))


```

```python
y_train_enc.value_counts() # three categories are now Label encoded
```

```python
# Let us use StandardScaler to scale the Result
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.set_output(transform='pandas')

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(X_train_scaled.sample(5))
```

# Let us fit the Logistic Regression Model Now

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000)

lr.fit(X_train_scaled , y_train_enc)

y_pred = lr.predict(X_test_scaled)
```

- This lr.predict gives us discrete values either 0,1 or 2

```python
from sklearn.metrics import mean_squared_error , accuracy_score , classification_report
from sklearn.metrics import confusion_matrix

print(f"root_mean_squared_error-> {np.sqrt(mean_squared_error(y_test_enc , y_pred))}\n")
print(f'Accuracy Score -> {accuracy_score(y_test_enc , y_pred)}\n')
print(f"Confusion matrix -> {confusion_matrix(y_test_enc , y_pred)}\n")
print(f"Classification Report -> {classification_report(y_test_enc , y_pred)}\n")
```

### Finding the prediction probabilities

```python
y_prob = lr.predict_proba(X_test_scaled) # only taking the 2nd array since second one has the data of positive predictions
```

```python
y_prob
```

```python
from sklearn.metrics import roc_auc_score

auc = roc_auc_score(
    y_test_enc,
    y_prob,
    multi_class='ovr'
)

print(auc)
```
