### ***We will learn different methods to Apply Hyperparameter tuning on the model and find the best parameters for the model***

```python
# import necessary modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split, KFold , StratifiedKFold ,cross_val_score
from sklearn.metrics import r2_score
```

```python
# load the dataset we want
df = pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv')

print(df.head())
```

```python
print(df.info())
```

### Fitting the Default Model With 5 fold cross validation

```python
# Prepare the data

X = df.iloc[:,:-1]
y = df.iloc[:,-1]

# Let us fit the KNeighborsRegressor Model default values
knn = KNeighborsRegressor(n_neighbors=3 , weights='uniform' , p=2)

kf =  KFold(n_splits = 5 , shuffle=True,random_state=1)

scores = cross_val_score(knn , X,y, cv = kf , scoring='r2')

print(f"Scores -> {scores}")
print(f"Best Score -> {scores.mean()}")

```

### Let us Now use  Hyperparameter tuning 

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2, random_state=42)

param_grid = {
    'n_neighbors' : [1,3,4,5,6],
    'weights':['uniform','distance'],
    'algorithm':['brute','ball_tree','kd_tree'],
    'p': [1,2]
}

from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(knn , param_grid , cv = kf , scoring='r2')

grid.fit(X_train,y_train)

print(grid.best_params_)

```

```python
pd.DataFrame(grid.cv_results_).sort_values(by='mean_test_score' , ascending=False)
```

```python
grid.best_estimator_
```

```python
grid.best_score_ # This is the mean score of the model with best parameters
```

```python
y_pred = grid.predict(X_test)

print(f"r2_score -> {r2_score(y_test , y_pred)}")
```

### GridSearchCV is a brute force algo, which fits all the pssible paramters combinations - Can take a very long time if there are high no. of models to train

### So we use RandomisedsearchCV sometimes

```python
from sklearn.model_selection import RandomizedSearchCV

grid = RandomizedSearchCV(knn ,param_grid , cv = kf , scoring='r2')

grid.fit(X_train , y_train)

print(grid.best_params_)
```

```python
pd.DataFrame(grid.cv_results_).sort_values(by='mean_test_score' , ascending=False)
```

- Here we see only 10 models were fit since by default n_iter = 10 , but we can specify our own also

```python
# Choosing our own number of iterations

from sklearn.model_selection import RandomizedSearchCV

grid = RandomizedSearchCV(knn ,param_grid ,n_iter = 30, cv = kf , scoring='r2')

grid.fit(X_train , y_train)

print(grid.best_params_)
```
