```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
```

```python
df = pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,:5] , df.iloc[:,-1] , test_size=.2 , random_state=42)
```

```python
bag = BaggingClassifier(max_features=3 , oob_score=True)
```

```python
bag.fit(X_train,y_train)
```

```python
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[9])
plt.show()
```

```python
from sklearn.metrics import accuracy_score
y_pred = bag.predict(X_test)

print(f'Acc Score -> {accuracy_score(y_test , y_pred)}')
```

```python
bag.oob_score_
```

```python
rf = RandomForestClassifier(max_features=3 , oob_score=True)
```

```python
rf.fit(X_train,y_train)
```

```python
plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
```

```python
y_pred = rf.predict(X_test)
print(f"Acc Score-> {accuracy_score(y_test,y_pred)}")

print(f'Acc From OOB -> {rf.oob_score_}')
```
